```python
import json, warnings, sklearn, numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
from pathlib import Path
from IPython.display import display, HTML
warnings.filterwarnings("ignore")

RANDOM_STATE = 42
OUT = Path("salidas_extra_v2"); OUT.mkdir(exist_ok=True)      # tablas auxiliares de esta versión
sns.set_theme(style="whitegrid")
plt.rcParams.update({"figure.dpi": 110, "axes.titleweight": "bold", "axes.titlesize": 12})
pd.set_option("display.max_columns", None)

FUENTE = "font-family:'Trebuchet MS','Segoe UI',sans-serif;"
def nota(titulo, cuerpo, color="#0e7c86", icono="🔎"):
    display(HTML(f'<div style="border-left:6px solid {color};background:{color}14;padding:12px 16px;margin:8px 0;'
                 f'border-radius:6px;{FUENTE}line-height:1.55"><b style="color:{color}">{icono} {titulo}</b><br>{cuerpo}</div>'))

def mostrar(d, alto=360):
    """Tabla completa dentro de un recuadro con scroll."""
    estilo = ("<style>.tabla{border-collapse:collapse;font-size:12px}.tabla th{position:sticky;top:0;background:#5c2250;"
              "color:white;padding:4px 8px}.tabla td{padding:3px 8px;border-bottom:1px solid #eadbe6;white-space:nowrap}</style>")
    cuerpo = d.to_html(max_rows=None, classes="tabla", border=0, na_rep="", float_format=lambda v: f"{v:.5g}")
    display(HTML(estilo + f'<div style="max-height:{alto}px;overflow:auto;border:2px solid #5c2250;border-radius:6px">{cuerpo}</div>'))

def guardar(d, nombre):
    d.to_csv(OUT / nombre, index=False, encoding="utf-8-sig")     # utf-8-sig: Excel abre bien los acentos

print("pandas", pd.__version__, "| scikit-learn", sklearn.__version__)
```

<div style="border-left:6px solid #1d4e89;background:#1d4e8914;padding:12px 16px;margin:8px 0;border-radius:6px;font-family:'Trebuchet MS','Segoe UI',sans-serif;line-height:1.55"><b style="color:#1d4e89">📘 Contexto</b><br><b>Lencería Lizzy</b> es un negocio familiar de ropa de la ciudad de Yacuiba que vende en el <b>Mercado Lourdes</b> y en la <b>Feria de Barrio Nuevo</b>, y también por catálogo en línea. <div style="background:linear-gradient(90deg,#4a1942,#8e3b7c 70%,#c06c9f);color:white;padding:14px 20px;border-radius:8px;margin-top:22px;font-family:'Trebuchet MS','Segoe UI',sans-serif;font-size:22px;font-weight:bold">🌸 2. Datos exportados de la base</div>

<div style="border-left:6px solid #1d4e89;background:#1d4e8914;padding:12px 16px;margin:8px 0;border-radius:6px;font-family:'Trebuchet MS','Segoe UI',sans-serif;line-height:1.55"><b style="color:#1d4e89">📘 Qué se hace y por qué</b><br>El sistema exporta cuatro archivos con <code>python manage.py exportar_dataset_v2</code>. Se excluyen los pedidos cancelados y los de prueba, y <b>no se exporta ningún dato de las clientas</b>. <ul><li><b>ventas_lineas.csv</b>: una fila por línea de pedido (es la base del modelo).</li><li><b>pedidos.csv</b>: una fila por pedido (sirve para análisis como el ticket promedio).</li><li><b>catalogo_productos.csv</b>: una fila por producto del catálogo.</li><li><b>metadatos_exportacion.json</b>: cuántas filas, pedidos, prendas y bolivianos declaró el exportador. Aquí se <b>verifica</b> que lo leído coincida con lo declarado.</li></ul></div>

```python
def buscar_carpeta():
    for c in ["datos/v2", "../datos/v2"]:
        if (Path(c) / "ventas_lineas.csv").exists():
            return Path(c)
    return None

CARPETA = buscar_carpeta()
if CARPETA is None:
    raise FileNotFoundError("No se encontró datos/v2/. En la laptop: python manage.py exportar_dataset_v2, subirlo con git y hacer git pull aquí.")

meta = json.loads((CARPETA / "metadatos_exportacion.json").read_text(encoding="utf-8"))
lineas = pd.read_csv(CARPETA / "ventas_lineas.csv", parse_dates=["fecha"])
pedidos = pd.read_csv(CARPETA / "pedidos.csv", parse_dates=["fecha"])
catalogo = pd.read_csv(CARPETA / "catalogo_productos.csv")

leido = {"lineas": len(lineas), "pedidos": lineas["pedido_id"].nunique(), "prendas": int(lineas["cantidad"].sum()),
         "subtotal_lineas_bs": round(float(lineas["subtotal"].sum()), 2), "dias_distintos": lineas["fecha"].nunique(),
         "fecha_min": str(lineas["fecha"].min().date()), "fecha_max": str(lineas["fecha"].max().date()),
         "productos_catalogo": len(catalogo)}

def igual(a, b):
    try: return abs(float(a) - float(b)) < 0.005
    except (TypeError, ValueError): return str(a) == str(b)

comp = pd.DataFrame({"Declarado por el exportador": {k: meta[k] for k in leido}, "Leído del CSV": leido})
comp["Coincide"] = ["✔" if igual(meta[k], leido[k]) else "✘" for k in leido]
assert (comp["Coincide"] == "✔").all(), "Lo leído no coincide con los metadatos: volver a exportar"
assert len(pedidos) == meta["pedidos"] and abs(pedidos["total_pedido"].sum() - meta["total_pedidos_bs"]) < 0.01, "pedidos.csv no coincide con los metadatos"
assert list(lineas.columns) == meta["columnas_lineas"], "Las columnas de ventas_lineas.csv no son las declaradas"
assert not meta["incluye_pruebas"] and (lineas["origen"] != "prueba").all(), "Hay pedidos de prueba: no deben entrenar el modelo"

print(f"Carpeta : {CARPETA.resolve()}")
print(f"Exportado el {meta['fecha_exportacion']} (versión {meta['version']})")
print(f"Orígenes presentes: {lineas['origen'].value_counts().to_dict()}   |   Pedidos por lugar: {meta['pedidos_por_lugar']}")
mostrar(comp, 330)
```

Si todas llevan ✔, los archivos están completos y no se dañaron al copiarlos con git. Si alguna lleva ✘, el notebook se detiene (<code>assert</code>) antes de entrenar con datos dudosos. Las cifras ya no están escritas a mano en el texto: se leen de los archivos, así que se actualizan solas al volver a exportar.</div>

```python
archivos = pd.DataFrame([{"Archivo": n, "Qué contiene": q, "Filas": len(d), "Columnas": d.shape[1],
                          "Tamaño (KB)": round((CARPETA / n).stat().st_size / 1024, 1)}
                         for n, q, d in [("ventas_lineas.csv", "una fila por línea de pedido", lineas),
                                         ("pedidos.csv", "una fila por pedido", pedidos),
                                         ("catalogo_productos.csv", "una fila por producto del catálogo", catalogo)]])
mostrar(archivos, 170)
```

<div style="background:linear-gradient(90deg,#4a1942,#8e3b7c 70%,#c06c9f);color:white;padding:14px 20px;border-radius:8px;margin-top:22px;font-family:'Trebuchet MS','Segoe UI',sans-serif;font-size:22px;font-weight:bold">🌸 3. Diccionario de variables: las 25 columnas de las ventas (y las de los otros archivos)</div>

<div style="border-left:6px solid #1d4e89;background:#1d4e8914;padding:12px 16px;margin:8px 0;border-radius:6px;font-family:'Trebuchet MS','Segoe UI',sans-serif;line-height:1.55"><b style="color:#1d4e89">📘 Qué se hace y por qué</b><br>Antes de usar cualquier columna hay que saber <b>qué significa</b>. Aquí se documenta cada una: descripción, tipo, un ejemplo real, cuántos valores distintos tiene y qué porcentaje está vacío. Algunas columnas llevan «actual» en el nombre porque reflejan el estado del producto <b>hoy</b>, no el día de la venta. Todo queda guardado en <code>salidas_extra_v2/diccionario_variables.csv</code>.</div>

```python
DIC_LINEAS = {
    "pedido_id": ("Número del pedido; un pedido puede tener varias líneas", "identificador"),
    "fecha": ("Fecha de la venta (AAAA-MM-DD)", "fecha"),
    "dia_semana": ("Día de la semana de la venta (sin acentos: miercoles, sabado)", "categórica"),
    "mes": ("Mes de la venta (1 a 12)", "numérica entera"),
    "semana_iso": ("Número de semana del año según ISO", "numérica entera"),
    "hora": ("Hora de la venta (0 a 23); en las ventas estimadas es reconstruida", "numérica entera"),
    "lugar_venta": ("Dónde se vendió: online, mercado_lourdes o feria_barrio_nuevo", "categórica"),
    "origen": ("Cómo se obtuvo el dato: real, estimado o prueba", "categórica"),
    "producto_id": ("Identificador del producto en el catálogo", "identificador"),
    "producto": ("Nombre del producto (puede repetirse por ofertas)", "texto"),
    "categoria": ("Categoría del catálogo (Mujer, Hombre, Otros)", "categórica"),
    "subcategoria": ("Subcategoría dentro de la categoría", "categórica"),
    "temporada": ("Temporada del producto en el catálogo (por ejemplo verano o todo_el_anio)", "categórica"),
    "talla": ("Talla vendida", "categórica"),
    "color": ("Color vendido (puede estar vacío si el producto no maneja color)", "categórica"),
    "cantidad": ("Unidades vendidas en la línea", "numérica entera"),
    "precio_unitario": ("Precio cobrado por unidad (Bs)", "numérica decimal"),
    "subtotal": ("cantidad × precio_unitario (Bs)", "numérica decimal"),
    "metodo_pago": ("Cómo pagó: efectivo o qr", "categórica"),
    "estado_pedido": ("Estado del pedido (entregado en las ventas importadas)", "categórica"),
    "estado_pago": ("Estado del pago (verificado en las ventas importadas)", "categórica"),
    "tipo_entrega": ("Cómo se entregó (tienda en las ventas importadas)", "categórica"),
    "precio_lista_actual": ("Precio de lista del producto HOY, no el del día de la venta (Bs)", "numérica decimal"),
    "en_oferta_actual": ("1 si el producto está en oferta HOY, 0 si no", "booleana (0/1)"),
    "descuento_actual_pct": ("Porcentaje de descuento del producto HOY", "numérica entera"),
}
DIC_PEDIDOS = {
    "pedido_id": ("Número del pedido", "identificador"), "fecha": ("Fecha de la venta", "fecha"),
    "dia_semana": ("Día de la semana", "categórica"), "hora": ("Hora de la venta (0 a 23)", "numérica entera"),
    "lugar_venta": ("Dónde se vendió", "categórica"), "origen": ("real, estimado o prueba", "categórica"),
    "tipo_entrega": ("Cómo se entregó", "categórica"), "estado_pedido": ("Estado del pedido", "categórica"),
    "metodo_pago": ("efectivo o qr", "categórica"), "estado_pago": ("Estado del pago", "categórica"),
    "n_lineas": ("Cuántas líneas tiene el pedido", "numérica entera"),
    "n_prendas": ("Cuántas prendas suma el pedido", "numérica entera"),
    "total_pedido": ("Total del pedido (Bs)", "numérica decimal"),
}
DIC_CATALOGO = {
    "producto_id": ("Identificador del producto", "identificador"), "producto": ("Nombre del producto", "texto"),
    "categoria": ("Categoría del catálogo", "categórica"), "subcategoria": ("Subcategoría", "categórica"),
    "temporada": ("Temporada del producto", "categórica"), "precio": ("Precio de lista actual (Bs)", "numérica decimal"),
    "en_oferta": ("1 si está en oferta", "booleana (0/1)"), "descuento_pct": ("Porcentaje de descuento", "numérica entera"),
    "activo": ("1 si el producto sigue en el catálogo", "booleana (0/1)"),
    "n_variantes": ("Combinaciones talla-color disponibles", "numérica entera"),
    "stock_total": ("Unidades en stock sumando todas las variantes", "numérica entera"),
}

def diccionario(df, dic, archivo):
    faltan = [c for c in df.columns if c not in dic]
    assert not faltan, f"Columnas sin descripción en {archivo}: {faltan}"
    filas = []
    for c in df.columns:
        v = df[c].dropna()
        ej = v.iloc[0] if len(v) else ""
        ej = ej.date() if isinstance(ej, pd.Timestamp) else ej
        filas.append({"N°": len(filas) + 1, "Columna": c, "Descripción": dic[c][0], "Tipo": dic[c][1], "Ejemplo": str(ej),
                      "Valores distintos": df[c].nunique(), "% vacíos": round(df[c].isna().mean() * 100, 1)})
    return pd.DataFrame(filas)

dic_lineas = diccionario(lineas, DIC_LINEAS, "ventas_lineas.csv")
dic_pedidos = diccionario(pedidos, DIC_PEDIDOS, "pedidos.csv")
dic_catalogo = diccionario(catalogo, DIC_CATALOGO, "catalogo_productos.csv")
guardar(pd.concat([d.assign(Archivo=a) for d, a in [(dic_lineas, "ventas_lineas.csv"), (dic_pedidos, "pedidos.csv"),
                                                    (dic_catalogo, "catalogo_productos.csv")]])[["Archivo", "N°", "Columna", "Descripción", "Tipo", "Ejemplo", "Valores distintos", "% vacíos"]],
        "diccionario_variables.csv")

display(HTML("<h4>ventas_lineas.csv: las 25 columnas</h4>")); mostrar(dic_lineas, 560)
display(HTML("<h4>pedidos.csv</h4>")); mostrar(dic_pedidos, 330)
display(HTML("<h4>catalogo_productos.csv</h4>")); mostrar(dic_catalogo, 330)
print(f"Variables documentadas: {len(dic_lineas)} + {len(dic_pedidos)} + {len(dic_catalogo)} = {len(dic_lineas) + len(dic_pedidos) + len(dic_catalogo)} (en 3 archivos)")
```

<div style="border-left:6px solid #1d4e89;background:#1d4e8914;padding:12px 16px;margin:8px 0;border-radius:6px;font-family:'Trebuchet MS','Segoe UI',sans-serif;line-height:1.55"><b style="color:#1d4e89">📘 Qué se hace y por qué</b><br>Cada una de las 25 columnas recibe una <b>decisión</b> y un <b>motivo</b>. Hay cuatro razones para no usar una columna en el modelo: <b>(a)</b> solo se conoce <b>tras la venta</b> (usarla sería fuga de información, porque el modelo debe pronosticar <i>antes</i> de vender); <b>(b)</b> es <b>constante</b> (vale lo mismo en todas las filas y no informa nada); <b>(c)</b> es un <b>identificador</b> o está <b>reemplazada</b> por otra variable que ya la resume; <b>(d)</b> es un <b>dato actual</b> (el estado del producto hoy, no el del día de la venta). Las columnas descartadas del modelo igual se usan en las gráficas descriptivas de la sección 9.</div>

```python
DECISION = {
    "pedido_id": ("Identificador", "Solo identifica el pedido; se usa para contar pedidos, no para predecir."),
    "fecha": ("USADA", "Base del modelo: de ella salen el día, el mes, la semana y la temporada de calor."),
    "dia_semana": ("USADA (derivada de fecha)", "Entra al modelo; se recalcula desde la fecha (el CSV la trae sin acentos)."),
    "mes": ("USADA (derivada de fecha)", "Entra al modelo; recoge la época del año."),
    "semana_iso": ("USADA (derivada de fecha)", "Entra al modelo; recoge la evolución dentro del año."),
    "hora": ("Solo se conoce tras la venta", "La hora existe una vez hecha la venta; además el modelo predice por día, no por hora."),
    "lugar_venta": ("USADA", "Entra al modelo: dónde se vende cambia lo que se vende."),
    "origen": ("Constante", "Todas las filas son 'estimado' (los pedidos de prueba se excluyen al exportar)."),
    "producto_id": ("Identificador", "Solo etiqueta al producto; no describe el fenómeno."),
    "producto": ("Transformada en tipo", "Se convierte en el tipo de prenda con un diccionario (sección 7)."),
    "categoria": ("Reemplazada por tipo", "El tipo de prenda ya la resume con más detalle."),
    "subcategoria": ("Reemplazada por tipo", "El tipo de prenda ya la resume con más detalle."),
    "temporada": ("Reemplazada por tipo", "El indicador de calor y el tipo ya recogen la temporada."),
    "talla": ("Solo se conoce tras la venta", "Describe la prenda ya vendida; no se sabe antes de vender."),
    "color": ("Solo se conoce tras la venta", "Describe la prenda ya vendida; no se sabe antes de vender."),
    "cantidad": ("Objetivo", "Se suma por día, lugar y tipo para obtener las unidades: lo que se predice."),
    "precio_unitario": ("Solo se conoce tras la venta", "Precio cobrado en la línea vendida."),
    "subtotal": ("Solo se conoce tras la venta", "Es cantidad × precio: contiene la respuesta (fuga de información)."),
    "metodo_pago": ("Solo se conoce tras la venta", "Se decide al pagar, después de la venta."),
    "estado_pedido": ("Constante", "Vale 'entregado' en todas las ventas importadas."),
    "estado_pago": ("Constante", "Vale 'verificado' en todas las ventas importadas."),
    "tipo_entrega": ("Constante", "Vale 'tienda' en todas las ventas importadas."),
    "precio_lista_actual": ("Dato actual, no histórico", "Es el precio del producto HOY, no el del día de la venta."),
    "en_oferta_actual": ("Dato actual, no histórico", "Indica si el producto está en oferta HOY."),
    "descuento_actual_pct": ("Dato actual, no histórico", "Es el descuento del producto HOY."),
}
assert set(DECISION) == set(lineas.columns), f"Columnas sin decisión o sobrantes: {set(lineas.columns) ^ set(DECISION)}"
USADAS = ("Objetivo", "USADA", "USADA (derivada de fecha)", "Transformada en tipo")

registro = pd.DataFrame([{"N°": i + 1, "Columna": c, "Decisión": DECISION[c][0], "Motivo": DECISION[c][1],
                          "Valores distintos": lineas[c].nunique(), "% vacíos": round(lineas[c].isna().mean() * 100, 1),
                          "¿Se usa?": "Sí" if DECISION[c][0] in USADAS else "No"} for i, c in enumerate(lineas.columns)])
guardar(registro, "registro_columnas.csv")

cambiaron = [c for c, (d, _) in DECISION.items() if d == "Constante" and lineas[c].nunique() > 1]
if cambiaron:
    nota("Revisar", f"Estas columnas se declararon constantes pero ya no lo son: <b>{', '.join(cambiaron)}</b>. Conviene revisar la decisión (por ejemplo, si ya hay ventas reales).", "#b8651b", "⚠️")
mostrar(registro, 620)
```

<div style="border-left:6px solid #1d4e89;background:#1d4e8914;padding:12px 16px;margin:8px 0;border-radius:6px;font-family:'Trebuchet MS','Segoe UI',sans-serif;line-height:1.55"><b style="color:#1d4e89">📘 Qué se hace y por qué</b><br>Resume el proceso: cuántos datos entran y cuántos quedan en cada paso. Las <b>filas</b> no bajan al descartar columnas; bajan después, cuando las líneas de venta se agrupan en la tabla diaria (sección 8). El último paso se agrega en la sección 10, cuando ya están definidas las variables finales.</div>

```python
def grafico_pasos(p):
    fig, ax = plt.subplots(1, 2, figsize=(13, 0.9 + 0.8 * len(p)))
    colores = ["#1d4e89", "#4a90d9", "#8e3b7c", "#c06c9f", "#b8651b"][:len(p)]
    for a, col in zip(ax, ["Filas", "Columnas"]):
        barras = a.barh(p["Paso del proceso"].str.slice(0, 44)[::-1], p[col][::-1], color=colores[::-1])
        for b in barras: a.text(b.get_width(), b.get_y() + b.get_height() / 2, f" {int(b.get_width()):,}", va="center")
        a.set_title(f"{col} en cada paso"); a.set_xlim(0, p[col].max() * 1.2)
    plt.tight_layout(); plt.show()

cols_usadas = registro.loc[registro["¿Se usa?"] == "Sí", "Columna"].tolist()
pasos = pd.DataFrame([("1. Archivo exportado (ventas_lineas.csv)", len(lineas), lineas.shape[1]),
                      ("2. Tras descartar columnas (sección 4)", len(lineas), len(cols_usadas))],
                     columns=["Paso del proceso", "Filas", "Columnas"])
mostrar(pasos, 150); grafico_pasos(pasos)
nota("Lectura automática", f"Se parte de <b>{len(lineas)}</b> líneas y <b>{lineas.shape[1]}</b> columnas. Se descartan <b>{lineas.shape[1] - len(cols_usadas)}</b> columnas "
     f"y quedan <b>{len(cols_usadas)}</b>: {', '.join(cols_usadas)}.", "#0b6e4f", "✅")
```

Cada fila de la primera tabla es una línea de pedido; la segunda resume cada pedido; la tercera es el catálogo actual. Se desplazan con las barras de cada recuadro.</div>

<div style="background:linear-gradient(90deg,#4a1942,#8e3b7c 70%,#c06c9f);color:white;padding:14px 20px;border-radius:8px;margin-top:22px;font-family:'Trebuchet MS','Segoe UI',sans-serif;font-size:22px;font-weight:bold">🌸 7. Tipo de prenda</div>

<div style="border-left:6px solid #1d4e89;background:#1d4e8914;padding:12px 16px;margin:8px 0;border-radius:6px;font-family:'Trebuchet MS','Segoe UI',sans-serif;line-height:1.55"><b style="color:#1d4e89">📘 Qué se hace y por qué</b><br>La base organiza el catálogo por categoría (Mujer, Hombre, Otros) y subcategoría, pero para predecir interesa el <b>tipo de prenda</b> tal como se compra y se vende: ropa interior de mujer, cacheteros, hombre, niño, ropa usada, ropa de baño y medias. Ese tipo <b>no existe en la base</b>: se define aquí con un diccionario producto → tipo. Un <code>assert</code> verifica que ningún producto <b>vendido</b> quede sin tipo, y se revisa también el catálogo completo.</div>

```python
MAPEO = {
    # ropa interior de mujer
    "Boyshort": "ropa_interior_mujer", "Calzón clasico": "ropa_interior_mujer", "Calzón faja": "ropa_interior_mujer",
    "Colalé": "ropa_interior_mujer", "Corpiño": "ropa_interior_mujer", "Culotte": "ropa_interior_mujer",
    "Fajas": "ropa_interior_mujer", "Hilo Dental": "ropa_interior_mujer", "Tanga": "ropa_interior_mujer",
    "Bandeau top sin tiros": "ropa_interior_mujer", "Corpiño sin armado": "ropa_interior_mujer",
    "Cubre pezón": "ropa_interior_mujer", "Tirantes": "ropa_interior_mujer", "Top clásico": "ropa_interior_mujer",
    # cacheteros
    "Cachetero Liso": "cacheteros", "Cachetero con diseño": "cacheteros",
    # hombre
    "Brief": "hombre", "Bóxer ajustado (Boxer Brief)": "hombre", "Bóxer suelto": "hombre", "Slip": "hombre",
    # niño
    "Bóxer infantil": "nino", "Calzocillo de niño": "nino", "Medias infantiles": "nino",
    # ropa usada (mujer y hombre)
    "Ropa usada": "ropa_usada",
    # ropa de baño
    "Bikini": "ropa_de_bano", "Biquini usado": "ropa_de_bano", "Maya usada": "ropa_de_bano", "Traje de baño usado": "ropa_de_bano",
    # medias y pañuelos
    "Medias": "medias", "Medias Antideslizantes": "medias", "Medias de marca": "medias", "Pañuelos": "medias",
}
COLORES = {"ropa_interior_mujer": "#c06c9f", "cacheteros": "#8e3b7c", "hombre": "#1d4e89", "nino": "#0e7c86",
           "ropa_usada": "#b8651b", "ropa_de_bano": "#2a9d8f", "medias": "#6a3d9a"}
LUG_COL = {"mercado_lourdes": "#c06c9f", "feria_barrio_nuevo": "#1d4e89", "online": "#0e7c86"}

lineas["tipo"] = lineas["producto"].map(MAPEO)
sin_tipo = lineas.loc[lineas["tipo"].isna(), "producto"].unique()
assert len(sin_tipo) == 0, f"Productos vendidos sin tipo asignado: {list(sin_tipo)}"

resumen_tipo = (lineas.groupby("tipo").agg(lineas=("cantidad", "size"), prendas=("cantidad", "sum"),
                primera_venta=("fecha", "min"), ultima_venta=("fecha", "max")).reset_index())
resumen_tipo["primera_venta"] = resumen_tipo["primera_venta"].dt.date
resumen_tipo["ultima_venta"] = resumen_tipo["ultima_venta"].dt.date
mostrar(resumen_tipo, 270)

catalogo["tipo"] = catalogo["producto"].map(MAPEO)
catalogo["vendido"] = catalogo["producto_id"].isin(set(lineas["producto_id"]))
sin_tipo_cat = sorted(catalogo.loc[catalogo["tipo"].isna(), "producto"].unique())
if sin_tipo_cat:
    nota("Productos del catálogo sin tipo", f"Estos productos del catálogo no están en el diccionario (no afectan al modelo porque no tienen ventas, pero hay que asignarles tipo si se venden): <b>{', '.join(sin_tipo_cat)}</b>.", "#b8651b", "⚠️")
else:
    nota("Catálogo completo", f"Los <b>{len(catalogo)}</b> productos del catálogo tienen tipo asignado.", "#0b6e4f", "✅")
tabla_cat = (catalogo.groupby("tipo").agg(productos=("producto_id", "size"), con_ventas=("vendido", "sum")).reset_index())
display(HTML("<h4>Productos del catálogo por tipo</h4>")); mostrar(tabla_cat, 250)
```

<div style="border-left:6px solid #0e7c86;background:#0e7c8614;padding:12px 16px;margin:8px 0;border-radius:6px;font-family:'Trebuchet MS','Segoe UI',sans-serif;line-height:1.55"><b style="color:#0e7c86">🔎 Cómo leer esto</b><br>La primera tabla muestra, por tipo, cuántas líneas y prendas se vendieron y entre qué fechas. Se ve que <b>ropa de baño</b> y <b>ropa usada</b> aparecen recién desde septiembre: antes no existían en el negocio, y eso condiciona cómo se arman los ceros en la sección siguiente. La segunda tabla cuenta cuántos productos del catálogo tiene cada tipo y cuántos llegaron a venderse.</div>

<div style="border-left:6px solid #b8651b;background:#b8651b14;padding:12px 16px;margin:8px 0;border-radius:6px;font-family:'Trebuchet MS','Segoe UI',sans-serif;line-height:1.55"><b style="color:#b8651b">⚠️ Decisiones del diccionario que conviene poder defender</b><br><ul><li><b>Pañuelos</b> va con <b>medias</b>: es la categoría «Otros» y tiene pocas ventas.</li><li><b>Medias infantiles</b> va con <b>niño</b>, no con medias.</li><li><b>Bikini</b> va con <b>ropa de baño</b>: es un traje de baño, no ropa interior.</li></ul></div>

<div style="background:linear-gradient(90deg,#4a1942,#8e3b7c 70%,#c06c9f);color:white;padding:14px 20px;border-radius:8px;margin-top:22px;font-family:'Trebuchet MS','Segoe UI',sans-serif;font-size:22px;font-weight:bold">🌸 8. Tabla diaria con ceros</div>

<div style="border-left:6px solid #1d4e89;background:#1d4e8914;padding:12px 16px;margin:8px 0;border-radius:6px;font-family:'Trebuchet MS','Segoe UI',sans-serif;line-height:1.55"><b style="color:#1d4e89">📘 Qué se hace y por qué</b><br>El modelo necesita <b>una fila por día, lugar y tipo de prenda</b>, incluyendo los casos donde <b>no se vendió nada</b> (unidades = 0). Si se dejaran solo las filas con ventas, el modelo creería que siempre se vende algo. Reglas para los ceros: <ul><li>Solo cuentan las <b>jornadas reales</b> (pares fecha + lugar donde hubo ventas): no se inventan días en que no se salió a vender.</li><li>Cada tipo entra a la tabla <b>desde su primera venta</b>: la ropa de baño no tiene ceros en agosto porque todavía no se vendía.</li></ul></div>

```python
unidades = (lineas.groupby(["fecha", "lugar_venta", "tipo"], as_index=False)["cantidad"].sum().rename(columns={"cantidad": "unidades"}))
inicio = lineas.groupby("tipo")["fecha"].min().rename("inicio").reset_index()
jornadas = lineas[["fecha", "lugar_venta"]].drop_duplicates()

grilla = jornadas.merge(inicio, how="cross")
grilla = grilla[grilla["fecha"] >= grilla["inicio"]].drop(columns="inicio")

diario = grilla.merge(unidades, on=["fecha", "lugar_venta", "tipo"], how="left")
diario["unidades"] = diario["unidades"].fillna(0).astype(int)
diario = diario.sort_values(["fecha", "lugar_venta", "tipo"]).reset_index(drop=True)

assert diario["unidades"].sum() == lineas["cantidad"].sum(), "Se perdieron unidades al armar la tabla diaria"
print(f"Filas de la tabla diaria: {len(diario)}  |  jornadas: {len(jornadas)}  |  tipos: {diario['tipo'].nunique()}")
print(f"Unidades totales: {diario['unidades'].sum()} (iguales a las del CSV)  |  filas con 0 unidades: {(diario['unidades'] == 0).sum()} "
      f"({(diario['unidades'] == 0).mean():.0%})")
print("\nInicio de cada tipo:"); print(inicio.assign(inicio=inicio["inicio"].dt.date).to_string(index=False))
mostrar(diario.groupby("tipo")["unidades"].agg(filas="count", total="sum", promedio="mean", maximo="max").round(2).reset_index(), 270)
```

<div style="border-left:6px solid #0e7c86;background:#0e7c8614;padding:12px 16px;margin:8px 0;border-radius:6px;font-family:'Trebuchet MS','Segoe UI',sans-serif;line-height:1.55"><b style="color:#0e7c86">🔎 Cómo leer esto</b><br>Cada fila de <code>diario</code> responde: <i>«¿cuántas unidades de este tipo se vendieron ese día en ese lugar?»</i>. La suma de unidades es idéntica a la del CSV (se verifica con <code>assert</code>). El <b>porcentaje de ceros</b> indica cuántas combinaciones día-tipo no tuvieron ventas: es lo normal en un negocio pequeño y lo que hace útil predecir.</div>

<div style="border-left:6px solid #1d4e89;background:#1d4e8914;padding:12px 16px;margin:8px 0;border-radius:6px;font-family:'Trebuchet MS','Segoe UI',sans-serif;line-height:1.55"><b style="color:#1d4e89">📘 Qué se hace y por qué</b><br>Se agregan a la tabla diaria las variables que salen de la <b>fecha</b>: día de la semana (con acentos), mes, semana ISO y el <b>indicador de temporada de calor</b>, que vale 1 desde el <b>22 de septiembre de 2026</b> (inicio de la primavera). Ese indicador es un <b>supuesto propio</b>, no un dato medido.</div>

```python
DIAS = ["lunes", "martes", "miércoles", "jueves", "viernes", "sábado", "domingo"]
FECHA_CALOR = pd.Timestamp("2026-09-22")
diario["dia_semana"] = diario["fecha"].dt.dayofweek.map(dict(enumerate(DIAS)))
diario["mes"] = diario["fecha"].dt.month
diario["semana_iso"] = diario["fecha"].dt.isocalendar().week.astype(int)
diario["calor"] = (diario["fecha"] >= FECHA_CALOR).astype(int)
mostrar(diario.assign(fecha=diario["fecha"].dt.date).head(12), 300)
```

<div style="border-left:6px solid #0e7c86;background:#0e7c8614;padding:12px 16px;margin:8px 0;border-radius:6px;font-family:'Trebuchet MS','Segoe UI',sans-serif;line-height:1.55"><b style="color:#0e7c86">🔎 Cómo leer esto</b><br>Las primeras filas de la tabla diaria ya con las variables derivadas de la fecha. Compara <code>dia_semana</code> con la del CSV original: aquí lleva acentos (miércoles, sábado) porque se recalcula desde la fecha.</div>

<div style="background:linear-gradient(90deg,#4a1942,#8e3b7c 70%,#c06c9f);color:white;padding:14px 20px;border-radius:8px;margin-top:22px;font-family:'Trebuchet MS','Segoe UI',sans-serif;font-size:22px;font-weight:bold">🌸 9. Análisis exploratorio: 12 gráficas para entender los datos</div>

<div style="border-left:6px solid #b8651b;background:#b8651b14;padding:12px 16px;margin:8px 0;border-radius:6px;font-family:'Trebuchet MS','Segoe UI',sans-serif;line-height:1.55"><b style="color:#b8651b">⚠️ Cómo leer todas las gráficas de esta sección</b><br>Las ventas son <b>estimadas</b>: lo que se ve en las gráficas refleja las <b>reglas con las que se reconstruyeron</b> (cuántas ventas por día, qué prendas en qué mes), no preferencias reales de las clientas. Sirven para entender cómo está armado el dataset y qué puede aprender el modelo, no para sacar conclusiones del negocio real. Las gráficas de talla, color, hora y método de pago son las más sensibles a esto.</div>

<div style="border-left:6px solid #1d4e89;background:#1d4e8914;padding:12px 16px;margin:8px 0;border-radius:6px;font-family:'Trebuchet MS','Segoe UI',sans-serif;line-height:1.55"><b style="color:#1d4e89">📘 Qué se hace y por qué</b><br><b>9.1 Indicadores generales.</b> Antes de las gráficas, seis cifras que resumen el dataset: cuántos pedidos y prendas hay, cuánto suman en bolivianos, cuánto vale un pedido típico y cuánto se vende por jornada.</div>

```python
n_ped = len(pedidos); n_pr = int(lineas["cantidad"].sum()); ingreso = float(pedidos["total_pedido"].sum())
n_jor = len(jornadas); und_jor = diario.groupby(["fecha", "lugar_venta"])["unidades"].sum().mean()
def tarjeta(t, v, c):
    return (f'<div style="flex:1;min-width:150px;background:{c}18;border-top:5px solid {c};border-radius:8px;padding:10px 14px;{FUENTE}">'
            f'<div style="font-size:12px;color:#555">{t}</div><div style="font-size:26px;font-weight:bold;color:{c}">{v}</div></div>')
display(HTML('<div style="display:flex;gap:10px;flex-wrap:wrap;margin:6px 0">' + "".join([
    tarjeta("Pedidos", f"{n_ped:,}", "#8e3b7c"), tarjeta("Prendas vendidas", f"{n_pr:,}", "#c06c9f"),
    tarjeta("Total vendido", f"Bs {ingreso:,.2f}", "#1d4e89"), tarjeta("Pedido promedio", f"Bs {pedidos['total_pedido'].mean():.2f}", "#0e7c86"),
    tarjeta("Prendas por pedido", f"{pedidos['n_prendas'].mean():.2f}", "#b8651b"), tarjeta("Unidades por jornada", f"{und_jor:.1f}", "#2a9d8f")]) + "</div>"))
```

<div style="border-left:6px solid #0e7c86;background:#0e7c8614;padding:12px 16px;margin:8px 0;border-radius:6px;font-family:'Trebuchet MS','Segoe UI',sans-serif;line-height:1.55"><b style="color:#0e7c86">🔎 Cómo leer esto</b><br>Todas las cifras se calculan de los archivos exportados. Un <b>pedido</b> es una venta completa (puede tener varias líneas); una <b>jornada</b> es un día en un lugar donde hubo ventas. «Unidades por jornada» es el promedio de prendas que se venden en una jornada: es el orden de magnitud que el modelo debe aprender a predecir.</div>

<div style="border-left:6px solid #1d4e89;background:#1d4e8914;padding:12px 16px;margin:8px 0;border-radius:6px;font-family:'Trebuchet MS','Segoe UI',sans-serif;line-height:1.55"><b style="color:#1d4e89">📘 Qué se hace y por qué</b><br><b>9.2 Qué tipos de prenda pesan más.</b> A la izquierda, las <b>unidades</b> vendidas por tipo; a la derecha, los <b>bolivianos</b> que generaron. Pueden diferir: un tipo con muchas unidades baratas aporta menos dinero que uno con pocas unidades caras.</div>

```python
tot = diario.groupby("tipo")["unidades"].sum().sort_values()
ing = lineas.groupby("tipo")["subtotal"].sum().reindex(tot.index)
fig, ax = plt.subplots(1, 2, figsize=(14, 4.2))
for a, serie, tit, fmt in [(ax[0], tot, "Unidades vendidas por tipo", "{:.0f}"), (ax[1], ing, "Bolivianos vendidos por tipo (Bs)", "{:,.0f}")]:
    b = a.barh(serie.index, serie.values, color=[COLORES[t] for t in serie.index])
    for x in b: a.text(x.get_width(), x.get_y() + x.get_height() / 2, " " + fmt.format(x.get_width()), va="center")
    a.set_title(tit); a.set_xlim(0, serie.max() * 1.15)
plt.tight_layout(); plt.show()
nota("Lectura automática", f"El tipo con más unidades es <b>{tot.index[-1]}</b> ({int(tot.iloc[-1])} prendas, {tot.iloc[-1] / tot.sum():.0%} del total) y el que más dinero genera es "
     f"<b>{ing.idxmax()}</b> (Bs {ing.max():,.2f}). El tipo con menos ventas es <b>{tot.index[0]}</b> ({int(tot.iloc[0])} prendas).")
```

<div style="border-left:6px solid #0e7c86;background:#0e7c8614;padding:12px 16px;margin:8px 0;border-radius:6px;font-family:'Trebuchet MS','Segoe UI',sans-serif;line-height:1.55"><b style="color:#0e7c86">🔎 Cómo leer esto</b><br>Si el orden de las barras cambia entre izquierda y derecha, el tipo de prenda más vendido no es el que más dinero deja. A la derecha, los bolivianos <b>acumulados</b> a lo largo del tiempo.</div>

```python
dia = diario.groupby(["fecha", "lugar_venta"])["unidades"].sum().reset_index()
fig, ax = plt.subplots(1, 2, figsize=(14, 4.2))
for lugar, s in dia.groupby("lugar_venta"):
    ax[0].plot(s["fecha"], s["unidades"], "o-", color=LUG_COL[lugar], label=lugar)
ax[0].axvline(FECHA_CALOR, color="gray", ls="--"); ax[0].text(FECHA_CALOR, ax[0].get_ylim()[1] * 0.95, " inicio del calor", color="gray", va="top")
ax[0].set_title("Unidades por jornada y lugar"); ax[0].legend(); ax[0].tick_params(axis="x", rotation=35)
acum = lineas.groupby("fecha")["subtotal"].sum().cumsum()
ax[1].step(acum.index, acum.values, where="post", color="#8e3b7c", lw=2)
ax[1].set_title("Bolivianos acumulados"); ax[1].set_ylabel("Bs"); ax[1].tick_params(axis="x", rotation=35)
plt.tight_layout(); plt.show()
por_lugar = dia.groupby("lugar_venta")["unidades"].agg(["mean", "count"])
nota("Lectura automática", " ".join(f"En <b>{l}</b> hubo {int(r['count'])} jornadas con {r['mean']:.1f} unidades en promedio." for l, r in por_lugar.iterrows()) +
     f" El total acumulado llega a <b>Bs {acum.iloc[-1]:,.2f}</b>.")
```

<div style="background:linear-gradient(90deg,#4a1942,#8e3b7c 70%,#c06c9f);color:white;padding:14px 20px;border-radius:8px;margin-top:22px;font-family:'Trebuchet MS','Segoe UI',sans-serif;font-size:22px;font-weight:bold">🌸 10. Variables del modelo y división sin fuga de datos</div>

<div style="border-left:6px solid #1d4e89;background:#1d4e8914;padding:12px 16px;margin:8px 0;border-radius:6px;font-family:'Trebuchet MS','Segoe UI',sans-serif;line-height:1.55"><b style="color:#1d4e89">📘 Qué se hace y por qué</b><br>Se definen las <b>6 variables de entrada</b> y la variable objetivo (<code>unidades</code>). Las categóricas (lugar, día y tipo) se transforman con <b>One-Hot</b>; el mes y la semana se <b>escalan</b>; el indicador de calor ya vale 0 o 1. Todo ese preprocesamiento va <b>dentro del Pipeline</b>, para que en cada partición aprenda solo con los datos de entrenamiento.</div>

```python
import joblib
from sklearn.model_selection import GroupShuffleSplit, GroupKFold, cross_validate, GridSearchCV
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.dummy import DummyRegressor
from sklearn.linear_model import Ridge
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.inspection import permutation_importance

CAT = ["lugar_venta", "dia_semana", "tipo"]
NUM = ["mes", "semana_iso"]
BIN = ["calor"]
FEATS = CAT + NUM + BIN
OBJETIVO = "unidades"
X, y = diario[FEATS], diario[OBJETIVO]
grupos = diario["fecha"].dt.strftime("%Y-%m-%d") + "|" + diario["lugar_venta"]    # un grupo = una jornada

tabla_vars = pd.DataFrame([
    ("lugar_venta", "categórica", "One-Hot", "Dónde se vende cambia lo que se vende."),
    ("dia_semana", "categórica", "One-Hot", "Hay días de más movimiento que otros."),
    ("tipo", "categórica", "One-Hot", "Cada tipo de prenda tiene su propia demanda."),
    ("mes", "numérica", "StandardScaler", "Recoge la época del año."),
    ("semana_iso", "numérica", "StandardScaler", "Recoge la evolución dentro del año."),
    ("calor", "binaria 0/1", "sin cambio", f"1 desde el {FECHA_CALOR.date()} (supuesto propio, no dato medido)."),
    ("unidades", "OBJETIVO", "(no se transforma)", "Lo que se quiere predecir: prendas vendidas por día, lugar y tipo.")],
    columns=["Variable", "Tipo", "Preprocesamiento", "Por qué entra"])
guardar(tabla_vars, "variables_modelo.csv")
mostrar(tabla_vars, 290)
preproc = ColumnTransformer([("cat", OneHotEncoder(handle_unknown="ignore"), CAT), ("num", StandardScaler(), NUM), ("bin", "passthrough", BIN)])

pasos.loc[len(pasos)] = ["3. Tabla diaria (día × lugar × tipo)", len(diario), len(FEATS) + 1]
guardar(pasos, "pasos_proceso.csv"); guardar(diario.assign(fecha=diario["fecha"].dt.date), "dataset_diario_v2.csv")
display(HTML("<h4>De cuántas filas y columnas se parte y a cuántas se llega</h4>")); mostrar(pasos, 170); grafico_pasos(pasos)
```

<div style="border-left:6px solid #0e7c86;background:#0e7c8614;padding:12px 16px;margin:8px 0;border-radius:6px;font-family:'Trebuchet MS','Segoe UI',sans-serif;line-height:1.55"><b style="color:#0e7c86">🔎 Cómo leer esto</b><br>La tabla lista las <b>6 variables de entrada</b> y el objetivo, y cómo se prepara cada una. El embudo final resume todo el proceso: de las <b>líneas y 25 columnas exportadas</b> se llega a una tabla de <b>una fila por día, lugar y tipo</b> con 6 variables y el objetivo. La tabla diaria queda guardada en <code>salidas_extra_v2/dataset_diario_v2.csv</code>.</div>

<div style="border-left:6px solid #1d4e89;background:#1d4e8914;padding:12px 16px;margin:8px 0;border-radius:6px;font-family:'Trebuchet MS','Segoe UI',sans-serif;line-height:1.55"><b style="color:#1d4e89">📘 Qué se hace y por qué</b><br><b>División sin fuga.</b> Las filas del mismo día se parecen mucho entre sí. Si se mezclaran al azar, un mismo día podría estar en entrenamiento y en prueba, y el modelo «vería» la respuesta. Por eso se divide <b>por jornadas</b> (<code>GroupShuffleSplit</code>), reservando el 20 % de los días para prueba. Se busca la primera semilla (desde 42) en que la prueba contiene los dos lugares y días con y sin calor.</div>

```python
semilla_ok = None
for semilla in range(RANDOM_STATE, RANDOM_STATE + 500):
    i_tr, i_te = next(GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=semilla).split(X, y, grupos))
    te = diario.iloc[i_te]
    if te["lugar_venta"].nunique() == 2 and te["calor"].nunique() == 2 and (te["unidades"] > 0).sum() >= 15:
        semilla_ok = semilla; break
assert semilla_ok is not None, "No se encontró una división con ambos lugares y días con y sin calor"
X_tr, X_te, y_tr, y_te = X.iloc[i_tr], X.iloc[i_te], y.iloc[i_tr], y.iloc[i_te]
g_tr = grupos.iloc[i_tr]
assert not set(grupos.iloc[i_tr]) & set(grupos.iloc[i_te]), "Hay jornadas en entrenamiento y en prueba a la vez"
dias_te = sorted(diario.iloc[i_te]["fecha"].dt.strftime("%Y-%m-%d").unique())

print(f"Semilla elegida: {semilla_ok}")
print(f"Entrenamiento: {len(X_tr)} filas / {g_tr.nunique()} jornadas   |   Prueba: {len(X_te)} filas / {grupos.iloc[i_te].nunique()} jornadas")
print("Días de prueba:", ", ".join(dias_te))
jor = diario[["fecha", "lugar_venta"]].drop_duplicates().copy()
jor["conjunto"] = np.where(jor["fecha"].dt.strftime("%Y-%m-%d").isin(dias_te), "Prueba", "Entrenamiento")
fig, ax = plt.subplots(figsize=(12, 2.8))
for (lugar, conj), s in jor.groupby(["lugar_venta", "conjunto"]):
    ax.scatter(s["fecha"], [lugar] * len(s), s=170 if conj == "Prueba" else 70, marker="s" if conj == "Prueba" else "o",
               color="#b32024" if conj == "Prueba" else LUG_COL[lugar], label=f"{conj}", edgecolor="white", zorder=3)
h, l = ax.get_legend_handles_labels(); u = dict(zip(l, h)); ax.legend(u.values(), u.keys(), loc="center right")
ax.set_title("Jornadas de entrenamiento (círculos) y de prueba (cuadrados rojos)"); ax.tick_params(axis="x", rotation=30); ax.grid(axis="y", alpha=0.2)
plt.tight_layout(); plt.show()
```

<div style="border-left:6px solid #0e7c86;background:#0e7c8614;padding:12px 16px;margin:8px 0;border-radius:6px;font-family:'Trebuchet MS','Segoe UI',sans-serif;line-height:1.55"><b style="color:#0e7c86">🔎 Cómo leer esto</b><br>Cada punto es una <b>jornada</b>. Los cuadrados rojos son los días reservados para la prueba final: el modelo <b>no los ve</b> mientras se entrena ni mientras se elige. Un <code>assert</code> comprueba que ninguna jornada esté en los dos conjuntos. Con solo unos pocos días de prueba, las métricas finales son <b>orientativas</b>, no definitivas.</div>

<div style="background:linear-gradient(90deg,#4a1942,#8e3b7c 70%,#c06c9f);color:white;padding:14px 20px;border-radius:8px;margin-top:22px;font-family:'Trebuchet MS','Segoe UI',sans-serif;font-size:22px;font-weight:bold">🌸 11. Comparación de modelos con validación cruzada</div>

<div style="border-left:6px solid #1d4e89;background:#1d4e8914;padding:12px 16px;margin:8px 0;border-radius:6px;font-family:'Trebuchet MS','Segoe UI',sans-serif;line-height:1.55"><b style="color:#1d4e89">📘 Qué se hace y por qué</b><br>Se comparan <b>cuatro modelos</b>, todos con el mismo preprocesamiento dentro del Pipeline: <ul><li><b>Referencia (promedio)</b>: predice siempre el promedio. Es el piso: si un modelo no lo supera, no aporta nada.</li><li><b>Ridge</b>: regresión lineal con penalización; simple y explicable.</li><li><b>Random Forest</b> y <b>Gradient Boosting</b>: árboles que capturan relaciones no lineales con pocos datos.</li></ul> La evaluación usa <b>GroupKFold de 5 particiones por jornada</b> sobre el entrenamiento (nunca sobre la prueba). Métricas: <b>RMSE</b> (error típico en unidades, castiga los errores grandes), <b>MAE</b> (error promedio) y <b>R²</b> (qué parte de la variación se explica; 0 = como el promedio).</div>

```python
def armar(modelo): return Pipeline([("prep", preproc), ("modelo", modelo)])
MODELOS = {
    "Referencia (promedio)": DummyRegressor(strategy="mean"),
    "Ridge": Ridge(alpha=1.0, random_state=RANDOM_STATE),
    "Random Forest": RandomForestRegressor(n_estimators=300, random_state=RANDOM_STATE, n_jobs=-1),
    "Gradient Boosting": GradientBoostingRegressor(random_state=RANDOM_STATE),
}
cv = GroupKFold(n_splits=5)
filas = []
for nombre, m in MODELOS.items():
    r = cross_validate(armar(m), X_tr, y_tr, groups=g_tr, cv=cv, n_jobs=-1,
                       scoring={"rmse": "neg_root_mean_squared_error", "mae": "neg_mean_absolute_error", "r2": "r2"})
    filas.append({"Modelo": nombre, "RMSE": -r["test_rmse"].mean(), "RMSE (desv.)": r["test_rmse"].std(),
                  "MAE": -r["test_mae"].mean(), "R²": r["test_r2"].mean()})
comp_cv = pd.DataFrame(filas).sort_values("RMSE").reset_index(drop=True)
guardar(comp_cv, "comparacion_modelos_cv.csv")
mostrar(comp_cv.round(3), 220)
fig, ax = plt.subplots(1, 3, figsize=(15, 3.8))
cm = ["#9a9a9a" if "Referencia" in n else "#8e3b7c" for n in comp_cv["Modelo"]]
ax[0].barh(comp_cv["Modelo"][::-1], comp_cv["RMSE"][::-1], xerr=comp_cv["RMSE (desv.)"][::-1], color=cm[::-1], capsize=4); ax[0].set_title("RMSE (menor es mejor)")
ax[1].barh(comp_cv["Modelo"][::-1], comp_cv["MAE"][::-1], color=cm[::-1]); ax[1].set_title("MAE (menor es mejor)"); ax[1].set_yticklabels([])
ax[2].barh(comp_cv["Modelo"][::-1], comp_cv["R²"][::-1], color=cm[::-1]); ax[2].set_title("R² (mayor es mejor)"); ax[2].set_yticklabels([]); ax[2].axvline(0, color="k", lw=0.8)
plt.tight_layout(); plt.show()
ref = comp_cv.loc[comp_cv["Modelo"] == "Referencia (promedio)", "RMSE"].iloc[0]; mej = comp_cv.iloc[0]
nota("Lectura automática", f"El mejor modelo sin optimizar es <b>{mej['Modelo']}</b> con RMSE <b>{mej['RMSE']:.3f}</b> (la referencia tiene {ref:.3f}): "
     f"mejora <b>{(1 - mej['RMSE'] / ref):.0%}</b> el error de predecir siempre el promedio. Su R² es <b>{mej['R²']:.3f}</b>.")
```

<div style="border-left:6px solid #0e7c86;background:#0e7c8614;padding:12px 16px;margin:8px 0;border-radius:6px;font-family:'Trebuchet MS','Segoe UI',sans-serif;line-height:1.55"><b style="color:#0e7c86">🔎 Cómo leer esto</b><br>Las barras grises son la <b>referencia</b>: el error de no saber nada. Cualquier modelo con barra de RMSE más corta que la gris aporta algo. La <b>línea negra</b> sobre las barras de RMSE es la desviación entre las 5 particiones: si es grande, el resultado depende mucho de qué días caen en cada una. R² cercano a 0 equivale a predecir el promedio; cuanto más alto, más variación se explica.</div>

<div style="background:linear-gradient(90deg,#4a1942,#8e3b7c 70%,#c06c9f);color:white;padding:14px 20px;border-radius:8px;margin-top:22px;font-family:'Trebuchet MS','Segoe UI',sans-serif;font-size:22px;font-weight:bold">🌸 12. Optimización con búsqueda de hiperparámetros</div>

<div style="border-left:6px solid #1d4e89;background:#1d4e8914;padding:12px 16px;margin:8px 0;border-radius:6px;font-family:'Trebuchet MS','Segoe UI',sans-serif;line-height:1.55"><b style="color:#1d4e89">📘 Qué se hace y por qué</b><br>Los modelos tienen «perillas» (hiperparámetros) que cambian su comportamiento. <code>GridSearchCV</code> prueba <b>todas las combinaciones</b> de una cuadrícula y evalúa cada una con la <b>misma validación cruzada por jornada</b>. Se elige la combinación con menor RMSE. La prueba final sigue sin tocarse.</div>

```python
GRILLAS = {
    "Ridge": (Ridge(random_state=RANDOM_STATE), {"modelo__alpha": [0.01, 0.1, 1, 3, 10, 30, 100]}),
    "Random Forest": (RandomForestRegressor(random_state=RANDOM_STATE, n_jobs=-1),
                      {"modelo__n_estimators": [200, 400], "modelo__max_depth": [3, 5, 8, None], "modelo__min_samples_leaf": [1, 3, 5]}),
    "Gradient Boosting": (GradientBoostingRegressor(random_state=RANDOM_STATE),
                          {"modelo__n_estimators": [100, 200], "modelo__learning_rate": [0.03, 0.05, 0.1],
                           "modelo__max_depth": [2, 3, 4]}),
}
busquedas, filas = {}, []
for nombre, (m, grilla_p) in GRILLAS.items():
    gs = GridSearchCV(armar(m), grilla_p, cv=cv, scoring="neg_root_mean_squared_error", n_jobs=-1)
    gs.fit(X_tr, y_tr, groups=g_tr)
    busquedas[nombre] = gs
    filas.append({"Modelo": nombre, "RMSE sin optimizar": float(comp_cv.loc[comp_cv["Modelo"] == nombre, "RMSE"].iloc[0]),
                  "RMSE optimizado": -gs.best_score_, "Combinaciones": len(gs.cv_results_["params"]),
                  "Mejores hiperparámetros": {k.replace("modelo__", ""): v for k, v in gs.best_params_.items()}})
opt = pd.DataFrame(filas).sort_values("RMSE optimizado").reset_index(drop=True)
guardar(opt.assign(**{"Mejores hiperparámetros": opt["Mejores hiperparámetros"].astype(str)}), "optimizacion_modelos.csv")
mostrar(opt.round(3), 220)
NOMBRE_MEJOR = opt.loc[0, "Modelo"]; MEJOR = busquedas[NOMBRE_MEJOR].best_estimator_; HIPERS = opt.loc[0, "Mejores hiperparámetros"]
nota("Modelo elegido", f"<b>{NOMBRE_MEJOR}</b> con hiperparámetros <code>{HIPERS}</code>. RMSE en validación cruzada: <b>{opt.loc[0, 'RMSE optimizado']:.3f}</b> "
     f"(antes de optimizar: {opt.loc[0, 'RMSE sin optimizar']:.3f}).", "#0b6e4f", "✅")
```

El modelo elegido es el de menor RMSE optimizado, y sus hiperparámetros quedan guardados.</div>

Izquierda: el RMSE de cada modelo antes y después de optimizar. Derecha: las 15 mejores combinaciones del modelo elegido; si muchas dan casi el mismo error, el resultado es estable y no depende de un ajuste afortunado.</div>

```python
fig, ax = plt.subplots(1, 2, figsize=(14, 4))
xs = np.arange(len(opt)); w = 0.38
ax[0].bar(xs - w / 2, opt["RMSE sin optimizar"], w, label="Sin optimizar", color="#cfa6c3")
ax[0].bar(xs + w / 2, opt["RMSE optimizado"], w, label="Optimizado", color="#8e3b7c")
ax[0].set_xticks(xs); ax[0].set_xticklabels(opt["Modelo"]); ax[0].set_ylabel("RMSE (unidades)"); ax[0].legend(); ax[0].set_title("Efecto de la optimización")
for k, v in enumerate(opt["RMSE optimizado"]): ax[0].text(k + w / 2, v, f"{v:.2f}", ha="center", va="bottom")
res_gs = pd.DataFrame(busquedas[NOMBRE_MEJOR].cv_results_); res_gs["RMSE"] = -res_gs["mean_test_score"]
top15 = res_gs.sort_values("RMSE").head(15).iloc[::-1]
etq = [", ".join(f"{k.replace('modelo__', '')}={v}" for k, v in p.items()) for p in top15["params"]]
ax[1].barh(range(len(top15)), top15["RMSE"], color="#8e3b7c"); ax[1].set_yticks(range(len(top15))); ax[1].set_yticklabels(etq, fontsize=8)
ax[1].set_xlim(top15["RMSE"].min() * 0.9, top15["RMSE"].max() * 1.02); ax[1].set_title(f"15 mejores combinaciones de {NOMBRE_MEJOR}"); ax[1].set_xlabel("RMSE (validación cruzada)")
plt.tight_layout(); plt.show()
rango = res_gs["RMSE"].max() - res_gs["RMSE"].min()
nota("Lectura automática", f"Optimizar bajó el RMSE de {NOMBRE_MEJOR} de <b>{opt.loc[0, 'RMSE sin optimizar']:.3f}</b> a <b>{opt.loc[0, 'RMSE optimizado']:.3f}</b> "
     f"({1 - opt.loc[0, 'RMSE optimizado'] / opt.loc[0, 'RMSE sin optimizar']:.0%} menos). Entre las {len(res_gs)} combinaciones probadas el RMSE va de {res_gs['RMSE'].min():.3f} a {res_gs['RMSE'].max():.3f}.")
```

En la derecha están las mejores configuraciones del modelo ganador; una <b>diferencia pequeña entre ellas</b> indica que el modelo no es sensible al ajuste fino.</div>

<div style="background:linear-gradient(90deg,#4a1942,#8e3b7c 70%,#c06c9f);color:white;padding:14px 20px;border-radius:8px;margin-top:22px;font-family:'Trebuchet MS','Segoe UI',sans-serif;font-size:22px;font-weight:bold">🌸 13. Evaluación final en días nunca vistos</div>

<div style="border-left:6px solid #1d4e89;background:#1d4e8914;padding:12px 16px;margin:8px 0;border-radius:6px;font-family:'Trebuchet MS','Segoe UI',sans-serif;line-height:1.55"><b style="color:#1d4e89">📘 Qué se hace y por qué</b><br>Es la <b>única vez</b> que se usan los días de prueba. Se predice con el modelo elegido y se comparan las predicciones con las ventas reales. Se calculan las mismas tres métricas (RMSE, MAE y R²) y se comparan con la <b>referencia</b> (predecir siempre el promedio). Las predicciones negativas se llevan a 0, porque no existen ventas negativas. Las gráficas muestran <b>dónde acierta y dónde se equivoca</b> el modelo.</div>

```python
def metricas(yt, yp):
    return {"RMSE": float(np.sqrt(mean_squared_error(yt, yp))), "MAE": float(mean_absolute_error(yt, yp)), "R²": float(r2_score(yt, yp))}
ref_modelo = armar(DummyRegressor(strategy="mean")).fit(X_tr, y_tr)
pred_te = np.clip(MEJOR.predict(X_te), 0, None)
pred_ref = ref_modelo.predict(X_te)
m_mej, m_ref = metricas(y_te, pred_te), metricas(y_te, pred_ref)
met = pd.DataFrame([{"Modelo": "Referencia (promedio)", **m_ref}, {"Modelo": f"{NOMBRE_MEJOR} (optimizado)", **m_mej}])
guardar(met, "metricas_test.csv"); mostrar(met.round(3), 150)
mejora = 1 - m_mej["RMSE"] / m_ref["RMSE"]
nota("Lectura automática", f"En {len(X_te)} filas de {len(dias_te)} días nunca vistos, el modelo tiene RMSE <b>{m_mej['RMSE']:.3f}</b>, MAE <b>{m_mej['MAE']:.3f}</b> y R² <b>{m_mej['R²']:.3f}</b>. "
     f"La referencia tiene RMSE {m_ref['RMSE']:.3f} y R² {m_ref['R²']:.3f}: el modelo reduce el error en <b>{mejora:.0%}</b>. "
     "Con tan pocos días de prueba, estas cifras son orientativas.", "#0b6e4f", "✅")
```

<div style="border-left:6px solid #0e7c86;background:#0e7c8614;padding:12px 16px;margin:8px 0;border-radius:6px;font-family:'Trebuchet MS','Segoe UI',sans-serif;line-height:1.55"><b style="color:#0e7c86">🔎 Cómo leer esto</b><br>El <b>RMSE</b> es el error típico en unidades de prenda: un RMSE de 1,7 significa que, en promedio, el modelo se equivoca por menos de 2 prendas por día, lugar y tipo. El <b>MAE</b> es el error promedio sin castigar los grandes. El <b>R²</b> indica qué parte de la variación se explica (0 = como el promedio, 1 = perfecto). La fila de la referencia es el piso: sirve para saber cuánto aporta el modelo.</div>

```python
ev = X_te.copy(); ev["real"] = y_te.values; ev["predicho"] = pred_te; ev["fecha"] = diario.iloc[i_te]["fecha"].values
ev["error"] = ev["predicho"] - ev["real"]
fig, ax = plt.subplots(1, 3, figsize=(16, 4.3))
ax[0].scatter(ev["real"], ev["predicho"], alpha=0.7, color="#8e3b7c", edgecolor="white")
mx = float(max(ev["real"].max(), ev["predicho"].max())) + 1
ax[0].plot([0, mx], [0, mx], "k--", lw=1); ax[0].set_xlabel("Unidades reales"); ax[0].set_ylabel("Unidades predichas"); ax[0].set_title("Real frente a predicho (cada punto = una fila)")
ax[1].hist(ev["error"], bins=15, color="#c06c9f", edgecolor="white"); ax[1].axvline(0, color="k", ls="--"); ax[1].set_title("Errores (predicho − real)"); ax[1].set_xlabel("Unidades"); ax[1].set_ylabel("Filas")
pt = ev.groupby("tipo")[["real", "predicho"]].sum().sort_values("real")
yy = np.arange(len(pt)); ax[2].barh(yy - 0.2, pt["real"], 0.4, label="Real", color="#1d4e89"); ax[2].barh(yy + 0.2, pt["predicho"], 0.4, label="Predicho", color="#c06c9f")
ax[2].set_yticks(yy); ax[2].set_yticklabels(pt.index); ax[2].legend(); ax[2].set_title("Total de unidades por tipo en la prueba")
plt.tight_layout(); plt.show()
nota("Lectura automática", f"El error medio es <b>{ev['error'].mean():+.2f}</b> unidades (cerca de 0 = sin sesgo). El <b>{(ev['error'].abs() <= 1).mean():.0%}</b> de las predicciones se equivoca en 1 unidad o menos, "
     f"y el error más grande es de <b>{ev['error'].abs().max():.1f}</b> unidades.")
```

<div style="border-left:6px solid #1d4e89;background:#1d4e8914;padding:12px 16px;margin:8px 0;border-radius:6px;font-family:'Trebuchet MS','Segoe UI',sans-serif;line-height:1.55"><b style="color:#1d4e89">📘 Qué se hace y por qué</b><br>Tres gráficas más para <b>ver dónde falla</b>: el total de cada <b>jornada de prueba</b> (real y predicho), el <b>error promedio por tipo</b> y los <b>residuos</b> frente a lo predicho (si forman una nube sin forma, el modelo no tiene un sesgo sistemático).</div>

<div style="border-left:6px solid #0e7c86;background:#0e7c8614;padding:12px 16px;margin:8px 0;border-radius:6px;font-family:'Trebuchet MS','Segoe UI',sans-serif;line-height:1.55"><b style="color:#0e7c86">🔎 Cómo leer esto</b><br><b>Izquierda:</b> si las barras azul y rosada de una jornada tienen casi la misma altura, el modelo acertó el total de ese día. <b>Centro:</b> los tipos con más ventas suelen tener errores mayores, porque hay más para equivocarse. <b>Derecha:</b> si los puntos se abren como un embudo hacia la derecha, el modelo se equivoca más cuando predice ventas altas.</div>

<div style="background:linear-gradient(90deg,#4a1942,#8e3b7c 70%,#c06c9f);color:white;padding:14px 20px;border-radius:8px;margin-top:22px;font-family:'Trebuchet MS','Segoe UI',sans-serif;font-size:22px;font-weight:bold">🌸 14. Importancia de las variables</div>

<div style="border-left:6px solid #1d4e89;background:#1d4e8914;padding:12px 16px;margin:8px 0;border-radius:6px;font-family:'Trebuchet MS','Segoe UI',sans-serif;line-height:1.55"><b style="color:#1d4e89">📘 Qué se hace y por qué</b><br>Para saber <b>qué aprendió el modelo</b> se usa la <b>importancia por permutación</b>: se desordenan los valores de una variable en los días de prueba y se mide cuánto <b>empeora el RMSE</b>. Si empeora mucho, la variable importa; si no cambia, el modelo no la usa; si el error <b>baja</b> (valor negativo), esa variable añade ruido.</div>

```python
pi = permutation_importance(MEJOR, X_te, y_te, scoring="neg_root_mean_squared_error", n_repeats=30, random_state=RANDOM_STATE, n_jobs=-1)
imp = (pd.DataFrame({"Variable": FEATS, "Aumento del RMSE": pi.importances_mean, "Desv.": pi.importances_std})
       .sort_values("Aumento del RMSE", ascending=False).reset_index(drop=True))
guardar(imp, "importancia_permutacion.csv"); mostrar(imp.round(3), 230)
fig, ax = plt.subplots(figsize=(9, 3.6)); r = imp.iloc[::-1]
ax.barh(r["Variable"], r["Aumento del RMSE"], xerr=r["Desv."], color=["#8e3b7c" if v > 0 else "#b32024" for v in r["Aumento del RMSE"]], capsize=3)
ax.axvline(0, color="k", lw=0.8); ax.set_title("Importancia por permutación (aumento del RMSE)"); ax.set_xlabel("Unidades de RMSE")
plt.tight_layout(); plt.show()
nota("Lectura automática", f"La variable más importante es <b>{imp.loc[0, 'Variable']}</b> (el RMSE sube {imp.loc[0, 'Aumento del RMSE']:.2f} al desordenarla) y la siguiente es <b>{imp.loc[1, 'Variable']}</b> ({imp.loc[1, 'Aumento del RMSE']:.2f}). "
     f"Variables con importancia casi nula o negativa: <b>{', '.join(imp.loc[imp['Aumento del RMSE'] < 0.02, 'Variable']) or 'ninguna'}</b>.")
```
